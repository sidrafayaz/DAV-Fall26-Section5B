```python
df = pd.read_csv('adult.csv')

print(df.isnull().sum())

df.replace('?', np.nan, inplace=True)

print(df.isnull().sum())
```

```python
#task 2
missing_percentages = (df.isnull().sum() / len(df)) * 100

missing_info = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage': missing_percentages})

missing_info = missing_info.sort_values(by='Missing Percentage', ascending=False)
print(missing_info[missing_info['Missing Count'] > 0])

#workclass and occupation have 5% missing percentage
```

```python
#task 3

print(df.columns)
df.sample(10)

for col in ['workclass', 'education', 'marital.status', 'occupation', 'relationship', 'race', 'sex', 'native.country']:
    df[col].fillna('Unknown', inplace=True)

print(df[['workclass', 'education', 'marital.status', 'occupation', 'relationship', 'race', 'sex', 'native.country']].isnull().sum())
#I replaced them with "Unknwon" as replacing them with the mode could skew training. Just because most people are Craft-repair doesn't mean anyone that hasn't put in their occupation is also the same


for col in ['age', 'fnlwgt', 'education.num', 'capital.gain', 'capital.loss', 'hours.per.week']:
  df[col].fillna(df[col].median(), inplace=True)

#I replaced them with the median as the median does not affect statistical analysis as much as mean
```

```python
#task 4

print(df.duplicated().sum())

allCols = df.columns.tolist()
allColsExceptIncome = []
for col in allCols:
  if col != 'income':
      allColsExceptIncome.append(col)

df.duplicated(subset=allColsExceptIncome).sum()

```

```python
#task 5

cols = ['education', 'marital.status', 'native.country']
for col in cols:
  print(df[col].unique())

#no inconsistencies that might appear due to whitespace
```

```python
#task 7
print(df['occupation'].value_counts())
#most common s prof-specialty
print('\n -----------------------------------------------------------')
print((df['sex'].value_counts() / len(df)) * 100)
print('\n -----------------------------------------------------------')
print((df['income'][df['income'] == '>50K'].value_counts() / len(df)) * 100)
print('\n -----------------------------------------------------------')
print((df['income'][df['income'] == '<=50K'].value_counts() / len(df)) * 100)



#income classes are imbalanced as rows with income <=50k is a lot more than >50K (24698 vs 7839)
```

```python
#task 9

import sqlite3 as sql3
import pandas as pd

# Create an in-memory SQLite database connection
conn = sql3.connect(':memory:')

# Load the pandas DataFrame into the SQLite database
# The table name will be 'adult_income'
df.to_sql('adult_income', conn, index=False, if_exists='replace')

sqlDf = pd.read_sql_query('select * from adult_income', conn)
print(sqlDf.shape)
print(df.shape)

#both shapes are the same
```
