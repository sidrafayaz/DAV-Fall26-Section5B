```python
import pandas as pd
import numpy as np
```

```python
#Task 1 — Fix Hidden Missing Values
dataset = pd.read_csv("adult.csv")
# dataset.info()
# dataset.isnull().sum()
colList = dataset.columns
for i in colList:
  dataset[i].replace('?', np.nan, inplace=True)
dataset.isnull().sum()
```

```python
# Task : 2  Quantify Missingness
null_percentage = (dataset.isnull().mean() * 100).sort_values(ascending=False)
null_percentage.head(1)
```

```python
# Task : 3: Handle Missing Values
for i in colList:
  if i == 'workclass' or 'occupation':
    dataset[i].replace(np.nan, 'Unknown')

for i in colList:
  if dataset[i].isin(['Unknown']).any():
    print("TRUE")
```

```python
# TASK : 4: Detect Duplicates
exact_duplicates = dataset.duplicated().sum()
print(f"Exact duplicate rows: {exact_duplicates}")

dup_ignoring_income = dataset.duplicated(subset=dataset.columns.drop('income')).sum()
print(f"Duplicates ignoring target (income) are: {dup_ignoring_income}")
```

```python
# TASK : 5: Fix Inconsistent Categorical Data
t5_list = ['native.country', 'marital.status', 'education']
print("==== BEFORE STRIPING ====")
for i in t5_list:
  print(dataset[i].unique())
  print()

print('\n')
print("==== AFTER STRIPING =====")
for i in t5_list:

  dataset[i] = dataset[i].str.strip()
  print(dataset[i].unique())
  print()
```

```python
# TASK : 6 : Descriptive Statistics
print("==============================================")
print(dataset.describe())
print(dataset.describe(include='object'))
print("==============================================")

print()
print()
print("==============================================")
print("Age:")
print("Mean:", dataset['age'].mean())
print("Median:", dataset['age'].median())
print("==============================================")

print()
print()
print("==============================================")
print("Hours per week:")
print("Mean:", dataset['hours.per.week'].mean())
print("Median:", dataset['hours.per.week'].median())
print("==============================================")
```

```python
# TASK : 7: Frequency Analysis
print("Most common occupation is: ", dataset['occupation'].value_counts().head(1).index[0])
male = dataset['sex']

print("Percentage of male:", ((dataset['sex'] == 'Male').sum() / len(dataset['sex'])) * 100)
print("Percentage of female: ", ((dataset['sex'] == 'Female').sum() / len(dataset['sex'])) * 100)

print("Percentage of <=50K:", ((dataset['income'] == '<=50K').sum() / len(dataset['income'])) * 100)
print("Percentage of >50K:", ((dataset['income'] == '>50K').sum() / len(dataset['income'])) * 100)


# there is a disparity, alot of people with less than 50k income
```

```python
# Task 8— Cross-Check a Data Quality Assumption
education_check = dataset.groupby('education')['education.num'].unique()
print(education_check.sort_values(ascending=False))
```

```python
#Task 9— SQL-Based Data Extraction
import pandas as pd
import sqlite3

original_df = pd.read_csv("adult.csv")

print("Original shape:", original_df.shape)
conn = sqlite3.connect("adult_income.db")

original_df.to_sql(
    "adult_income",
    conn,
    if_exists="replace",
    index=False
)
query = """
SELECT *
FROM adult_income
WHERE age > 30
"""

sql_df = pd.read_sql_query(query, conn)

print(sql_df.head())

sql_df = pd.read_sql_query("select count(*) from adult_income", conn)
sql_df
```
