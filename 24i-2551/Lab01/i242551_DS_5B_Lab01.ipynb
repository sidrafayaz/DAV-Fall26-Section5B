**task3**

```python
#for categorical columns with missing values, imputing with the mode is a simple strategy to preserve data integrity and avoid losing observations.
df['workclass'] = df['workclass'].fillna(df['workclass'].mode()[0])
df['occupation'] = df['occupation'].fillna(df['occupation'].mode()[0])
df['native.country'] = df['native.country'].fillna(df['native.country'].mode()[0])
display(df.isnull().sum())
```

```python
#for numerical, imputing with the mean is better for continuous data
#bcz it preserves the overall distribution and does not introduce new biases if the data is symmetrically distributed.
for col in df.select_dtypes(include=np.number).columns:
     if df[col].isnull().any():
         df[col].fillna(df[col].mean(), inplace=True)

df.isnull().sum()
```

**task 4**

```python
dupes = df.duplicated().sum()
print(dupes)
target_col = 'income'
not_target = [col for col in df.columns if col != target_col]
fDUPES= df.duplicated(subset=not_target).sum()
print(f"Duplicates ignoring target column: {fDUPES}")
df.drop_duplicates(inplace=True)
```
