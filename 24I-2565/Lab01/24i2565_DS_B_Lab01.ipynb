```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
```

```python
df = pd.read_csv("adult.csv")

df.head()
```

## Initial Data Inspection

```python
print(df.shape)
print(df.columns)
df.info()
```

## Task 1 — Fix Hidden Missing Values

```python
print(df.isnull().sum())

df.replace("?", np.nan, inplace=True)

print(df.isnull().sum())
```

## Task 2 — Quantify Missing Values

```python
missing_count = df.isnull().sum()
missing_percent = df.isnull().mean() * 100

missing = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing %": missing_percent
})

missing = missing.sort_values("Missing Count", ascending=False)

print(missing)
```

## Task 3 — Handle Missing Values

```python
df["workclass"] = df["workclass"].fillna("Unknown")
df["occupation"] = df["occupation"].fillna("Unknown")
df["native.country"] = df["native.country"].fillna("Unknown")

print(df.isnull().sum())
```

For workclass, occupation and native-country, missing values were replaced with `Unknown` instead of deleting the rows. These are categorical columns, so using an Unknown category keeps more data.

The numeric columns were checked for missing values, and no numeric imputation was needed.

## Task 4 — Detect Duplicates

```python
print("Exact duplicate rows:", df.duplicated().sum())
```

```python
columns_without_income = df.columns.drop("income")

print("Duplicates without income:",
      df.duplicated(subset=columns_without_income).sum())
```

Exact duplicate rows were checked using `duplicated()`. Duplicates were also checked while ignoring the income label because two different people can have the same attributes but different income outcomes.

## Task 5 — Fix Inconsistent Categorical Data

```python
print("Education:")
print(df["education"].unique())

print("\nMarital Status:")
print(df["marital.status"].unique())

print("\nNative Country:")
print(df["native.country"].unique())
```

```python
df["education"] = df["education"].str.strip()
df["marital.status"] = df["marital.status"].str.strip()
df["native.country"] = df["native.country"].str.strip()

print(df["education"].unique())
print(df["marital.status"].unique())
print(df["native.country"].unique())
```

## Task 6 — Descriptive Statistics

```python
df.describe()
```

```python
df.describe(include="object")
```

```python
print("Age Mean:", df["age"].mean())
print("Age Median:", df["age"].median())

print("Hours per Week Mean:", df["hours.per.week"].mean())
print("Hours per Week Median:", df["hours.per.week"].median())
```

The mean and median of age are fairly close. The same is true for hours-per-week, so they do not differ by a very large amount.

## Task 7 — Frequency Analysis

```python
print(df["occupation"].value_counts())

print("\nMost common occupation:")
print(df["occupation"].value_counts().idxmax())
```

```python
sex_percentage = df["sex"].value_counts(normalize=True) * 100
print(sex_percentage)
```

```python
income_percentage = df["income"].value_counts(normalize=True) * 100
print(income_percentage)
```

The income classes are imbalanced because the `<=50K` class has more records than the `>50K` class.

## Task 8 — Check Education and Education-Num

```python
education_check = df.groupby("education")["education.num"].unique()

print(education_check)
```

```python
for education, numbers in education_check.items():
    if len(numbers) > 1:
        print(education, numbers)
```

If the cell above prints nothing, each education level has a consistent education-num value and no inconsistencies were found.

## Task 9 — SQL-Based Data Extraction

```python
conn = sqlite3.connect("adult_income.db")

df.to_sql("adult_income", conn, if_exists="replace", index=False)

print("Database created")
```

```python
query = """
SELECT *
FROM adult_income
WHERE age > 30
"""

sql_df = pd.read_sql_query(query, conn)

sql_df.head()
```

```python
print("Original shape:", df.shape)
print("SQL result shape:", sql_df.shape)
```

```python
print("People with age > 30 in original data:",
      (df["age"] > 30).sum())

print("Rows returned by SQL:",
      len(sql_df))
```

```python
query2 = """
SELECT age, education, occupation, income
FROM adult_income
WHERE "hours.per.week" > 40
"""

sql_df2 = pd.read_sql_query(query2, conn)

sql_df2.head()
```

```python
conn.close()
```

## Task 10 — Data Profiling Summary Report

### Dataset Overview

The UCI Adult Income dataset contains information about people such as age, education, occupation, marital status, working hours and other attributes. The target column is income, which shows whether a person's income is `<=50K` or `>50K`.

### Data Quality Issues

Missing values were hidden using `?` and were changed to proper NaN values. Missing data was mainly found in workclass, occupation and native-country. Extra whitespace was also checked in categorical columns. Duplicate rows were checked both with and without the income label.

### Key Observations

The frequency analysis shows that some occupations are more common than others. The sex distribution contains more male records than female records. The income classes are imbalanced, with the `<=50K` class being larger.

The mean and median of age and hours-per-week are fairly close.

### Cleaning Decisions

Missing categorical values in workclass, occupation and native-country were replaced with `Unknown` instead of deleting the rows. Extra whitespace was removed from categorical values. Duplicate rows were checked but not automatically removed.
