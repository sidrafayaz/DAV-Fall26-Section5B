```python
import pandas as p
import numpy as n

df=p.read_csv("adult.csv")
print("First five rows:", df.head(5))
print("Info:", df.info())

print("Initial missing vals\n", df.isnull().sum())
df.replace("?", n.nan, inplace=True)
print("Missing vals\n", df.isnull().sum())
```

```python
missing=df.isnull().sum()
percentage=(missing/len(df))*100

table=p.DataFrame({"Missing count":missing, "Missing percentage":percentage})

table=table.sort_values(by="Missing count",ascending=False)

print(table)
```

```python
#workclass has 1800 missing values and dropping might remove useful data. so impute rather than replace with most common workclass
#occupation also has 1800 and impute for the same reason

df["workclass"] = df["workclass"].fillna("Unknown")
df["occupation"] = df["occupation"].fillna("Unknown")

df
```

```python
print("Duplicate rows\n", df.duplicated().sum())

print("Duplicates after ignoring target\n", df.drop(columns=["income"]).duplicated().sum())

#that one record is not duplicate since income can differ while having same details
```

# Data Profiling Summary Report

## 1. Dataset Overview

The Adult Income dataset contains 32,561 rows and 15 columns. It represents demographic, education, employment, and financial information about individuals, with the target variable `income` indicating whether a person's annual income is `<=50K` or `>50K`.

## 2. Data Quality Issues Found

- Missing values: Missing values were initially represented by `"?"`, so they were replaced with `NaN`.
- Workclass: 1,836 missing values (5.64%).
- Occupation: 1,843 missing values (5.66%).
- Native-country: 583 missing values (1.79%).
- All other columns had 0 missing values.
- Duplicates: Exact duplicate rows were checked using `duplicated().sum()`. Duplicates ignoring the target column were also checked. Records with identical features but different income labels were not treated as duplicates because they can represent valid different individuals.
- Inconsistent categorical values: Leading/trailing whitespace was identified as a potential inconsistency in categorical columns and was removed using `.str.strip()`.
- Education consistency: `education` and `education.num` were checked using `groupby()` and `.unique()`. Each education level mapped to a single education number, so no inconsistencies were found.

## 3. Key Observations

- The mean age is 38.58 years, while the median is 37 years, indicating a slight right skew.
- `hours.per.week` has a mean of 40.44 hours and a median of 40 hours, so there is very little difference between them.
- Prof-specialty is the most common occupation, with 4,140 records.
- The sex distribution is 66.92% Male and 33.08% Female.
- The income classes are imbalanced:
  - `<=50K`: 75.92%
  - `>50K`: 24.08%
- Therefore, the `<=50K` class is the clear majority class and class imbalance should be considered during machine-learning model evaluation.

## 4. Cleaning Decisions Made

- Replaced all `"?"` values with proper `NaN` values so that missing data could be detected correctly.
- Missing `workclass` and `occupation` values were imputed with `"Unknown"` rather than dropping the rows, because these columns contain about 5–6% missing values and the remaining information in those records is still useful.
- No numeric columns required imputation because there were no missing numeric values.
- Leading/trailing whitespace was removed from categorical columns using `.str.strip()` to ensure consistent category names.
- Exact duplicate rows were considered duplicates because all their attributes and target values were identical. Rows that were identical except for `income` were retained because different income outcomes can be valid observations.
