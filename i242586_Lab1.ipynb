```python
#task 1
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\Rayhab\Downloads\adult.csv")
print(df.head())
print(df.info())

print("Missing values before replacement=")
print(df.isnull().sum())
df.replace("?", np.nan, inplace=True)
print("\nMissing values after replacement:")
print(df.isnull().sum())
```

```python
#task2
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_data = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage': missing_percentage
})
missing_data = missing_data.sort_values(
    by='Missing Count',
    ascending=False
)
print(missing_data)
```

```python
#task 3
# workclass 
# Use Unknown instead of the mode. This preserves all rows and avoids assuming that people with missing workclass information belong to the most common workclass.
df['workclass'].fillna('Unknown', inplace=True)
#occupation 
#Use Unknown as well. Since occupation is categorical and the missingness may not be random, assigning the mode could introduce incorrect assumptions.
df['occupation'].fillna('Unknown', inplace=True)
#there are no missing values in numeric columns
```

```python
#task 4
# Exact duplicates
print("Exact duplicates:", df.duplicated().sum())

# Duplicates ignoring target
feature_columns = df.columns.drop('income')
print("Duplicates ignoring target:",
      df.duplicated(subset=feature_columns).sum())
#rows that are identical in their features but have different target values should generally be kept, because they may represent different individuals with different outcomes.
```

```python
#task 5
print("Education:")
print(df['education'].unique())

print("\nMarital Status:")
print(df['marital.status'].unique())

print("\nNative Country:")
print(df['native.country'].unique())

#just for practice
df['education'] = df['education'].str.strip()
df['marital.status'] = df['marital.status'].str.strip()
df['native.country'] = df['native.country'].str.strip()
```
