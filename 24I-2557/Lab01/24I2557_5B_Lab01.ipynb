```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('adult.csv')

print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")
print("Data Types & Summary:")
df.info()
print("\nFirst 5 rows:")
print(df.head())
```

```python
# Task 1

print("Missing values count before handling hidden markers:")
print(df.isnull().sum())

df.replace(['?', ' ?'], np.nan, inplace=True)

print("\nMissing values count after replacing '?' with np.nan:")
print(df.isnull().sum())
```

```python
# Task 2

missing_count = df.isnull().sum()
missing_pct = df.isnull().mean() * 100

missing_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage (%)': missing_pct.round(2)
})

missing_cols = missing_summary[missing_summary['Missing Count'] > 0].sort_values(
    by='Missing Count', ascending=False
)

print("Summary of Missing Data")
print(missing_cols)
```

```python
# Task 3

categorical_cols = ['workclass', 'occupation', 'native.country']
for col in categorical_cols:
    df[col] = df[col].fillna('Unknown')

print(f"Remaining null values across entire dataset: {df.isnull().sum().sum()}")
```

```python
# Task 4

exact_duplicates = df.duplicated().sum()
print(f"Exact duplicate rows: {exact_duplicates}")

feature_cols = [c for c in df.columns if c != 'income']
feature_duplicates = df.duplicated(subset=feature_cols).sum()
print(f"Duplicates ignoring income label: {feature_duplicates}")

duplicate_samples = df[df.duplicated(keep=False)].sort_values(by=list(df.columns[:3])).head(6)
print("\nSample Duplicate Rows:")
print(duplicate_samples[['age', 'workclass', 'education', 'occupation', 'income']])
```

```python
initial_rows = len(df)
df.drop_duplicates(inplace=True)
print(f"Removed {initial_rows - len(df)} duplicate rows. New shape: {df.shape}")
```

```python
# Task 5

object_cols = df.select_dtypes(include='object').columns

for col in object_cols:
    df[col] = df[col].astype(str).str.strip()

print("Distinct categories after stripping whitespace:")
print(f"\n1. Unique 'education' levels ({df['education'].nunique()}):")
print(df['education'].unique())

print(f"\n2. Unique 'marital.status' categories ({df['marital.status'].nunique()}):")
print(df['marital.status'].unique())

print(f"\n3. Unique 'native.country' values ({df['native.country'].nunique()}):")
print(df['native.country'].unique()[:10], "... (truncated)")
```

```python
# Task 6

print("Numerical Summary Statistics")
print(df.describe())
```

```python
print("Categorical Summary Statistics")
print(df.describe(include='object'))
```

```python
age_mean = df['age'].mean()
age_median = df['age'].median()

hours_mean = df['hours.per.week'].mean()
hours_median = df['hours.per.week'].median()

stats_comparison = pd.DataFrame({
    'Metric': ['Mean', 'Median', 'Difference (Mean - Median)'],
    'Age (years)': [round(age_mean, 2), round(age_median, 2), round(age_mean - age_median, 2)],
    'Hours per Week': [round(hours_mean, 2), round(hours_median, 2), round(hours_mean - hours_median, 2)]
})

print("Mean vs Median Comparison")
print(stats_comparison.to_string(index=False))
```

```python
# Task 7
occ_counts = df['occupation'].value_counts()
most_common_occ = occ_counts.index[0] if occ_counts.index[0] != 'Unknown' else occ_counts.index[1]
print(f"(a) Most common known occupation: {most_common_occ} ({occ_counts[most_common_occ]} occurrences)")
print("\nTop 5 Occupations:")
print(occ_counts.head(5))

sex_dist = df['sex'].value_counts(normalize=True) * 100
print(f"\n(b) Sex Distribution (%):\n{sex_dist.round(2)}")

income_dist = df['income'].value_counts(normalize=True) * 100
print(f"\n(c) Income Distribution (%):\n{income_dist.round(2)}")
```

```python
# Task 8
edu_to_num = df.groupby('education')['education.num'].unique()
print("Education Level -> Unique Numerical Values:")
for edu, nums in edu_to_num.items():
    print(f"  {edu:<15} -> {list(nums)}")

num_to_edu = df.groupby('education.num')['education'].unique()
is_one_to_one = all(len(nums) == 1 for nums in edu_to_num) and all(len(edus) == 1 for edus in num_to_edu)

print(f"\nIs education == education.num a strict 1-to-1 bijection? {is_one_to_one}")
```

```python
# Task 9
import sqlite3

connect = sqlite3.connect('adult_income.db')

df.to_sql('adult_income', connect, if_exists='replace', index=False)

query = "SELECT * FROM adult_income WHERE age > 30;"

df_sql = pd.read_sql_query(query, connect)

df_sql.head()
```

# Task 10 – Data Profiling Summary Report

## 1. Dataset Overview

- **Source:** UCI Adult Income Dataset (1994 US Census database), extracted by Barry Becker.
- **Purpose:** Predict whether a person's annual income is more than $50,000.
- **Rows:** 32,537
- **Columns:** 15
- **Features:** 6 numerical and 9 categorical.

## 2. Data Quality Issues

| Issue | Columns | Count / Percentage | Details |
|---|---|---|---|
| **Missing Values** | `occupation`, `workclass`, `native.country` | occupation: 1,843 (5.66%)<br>workclass: 1,836 (5.64%)<br>native.country: 583 (1.79%) | Missing values were represented by `?` instead of NaN. |
| **Duplicate Rows** | All columns | 24 (0.07%) | Some rows were exactly duplicated. |
| **Extra Whitespace** | Text columns | Multiple | Some category values contained leading or trailing spaces. |
| **Redundant Features** | `education`, `education.num` | 32,561 rows (100%) | Both columns represent the same education information in different formats. |

## 3. Key Observations

- **Income:** 75.91% earn **≤ $50K**, while 24.09% earn **> $50K**.
- This gives an income ratio of about **3.15 : 1**, so accuracy alone may not be a good evaluation metric.
- **Gender:** 66.92% are Male and 33.08% are Female.
- **Work Hours:** The median is **40 hours per week**, while the mean is **40.44 hours**.
- **Top Occupations:** `Prof-specialty` (4,136) and `Craft-repair` (4,094) are the most common occupations.
- `Exec-managerial` and `Prof-specialty` have the highest proportions of people earning **> $50K**.

## 4. Data Cleaning

1. Replaced missing values in `workclass`, `occupation`, and `native.country` with **`Unknown`**.
2. This preserved **2,399 rows** that would otherwise have been removed.
3. Removed **24 duplicate rows**.
4. The final dataset contains **32,537 rows**.
5. Used `.str.strip()` to remove extra spaces from categorical values.
6. Verified the data using **SQLite** and confirmed that the loaded data matched the original Pandas dataset in rows and columns.
