**Dataset Overview**
The dataset contains employment-related information about people, 
with income as the target variable. It contained 32561 rows and 15 columns.
After cleaning, the dataset contained 32537 rows.

**Data Quality Issues Found**
The dataset had missing or unknown values represented by '?' in some categorical columns
There were also 24 exact duplicate rows. When checking duplicates while ignoring the 
income column, 25 duplicates were found, indicating one additional case where identical features may have different income

**Key Observations**
The average age was 38.5 years and median age was 37. The median and average working hours were around 40 hours per week.
Prof-specialty was the most common occupation. The dataset contained around
66.9% males and 33.1% females
The income classes were imbalanced, with approximately 75.9% of records belonging to 
the <=50K class and 24.1% belonging to the >50K class.
No inconsistency was found between education and education.num, as each education 
level mapped to the same numerical value.

**Cleaning Decisions Made**

Missing or unknown categorical values were handled by replacing '?' with Unknown
The 24 exact duplicate rows were removed. Records with identical features but different 
income values were retained, because they may represent different individuals with similar attributes.
The cleaned dataset was loaded into a local sqLite database and extracted using SQL
